```python
df['date'] =  pd.to_datetime(df['date'])
df['year'] = df['date'].dt.year
df_1 = (df.groupby(['city', 'year'])['no_stations'].max().reset_index().copy() )
```

```python
import numpy as np
bins = [0, 1, 2, 3, 4, 10, 20, np.inf]

labels = [
    '1s',
    '2s',
    '3s',
    '4s',
    '5-10s',
    '10-20s',
    '20s+'
]

df_1['#stations/city'] = pd.cut(
    df_1['no_stations'],
    bins=bins,
    labels=labels
)
df_1
```

```python

city_type_yearly_counts = (
    df_1.groupby(
        ['year', '#stations/city'],
        observed=False
    )['city']
    .nunique()
)
```
