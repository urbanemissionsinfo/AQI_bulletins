```python
df['date'] =  pd.to_datetime(df['date'])
df['year'] = df['date'].dt.year
df_1 = (df.groupby(['city', 'year'])['no_stations'].max().reset_index().copy() )

## MODE LOGIC
def get_mode(series):
    modes=series.mode()
    return modes.max()

df_1 = (df.groupby(['city', 'year'])['no_stations'].agg(get_mode).reset_index().copy() )
```
